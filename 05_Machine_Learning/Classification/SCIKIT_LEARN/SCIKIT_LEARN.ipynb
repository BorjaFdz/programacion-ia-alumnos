## Creamos una instancia del modelo y fijamos los valores de los hiperparámetros

# DIVIDIR LOS DATOS

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,          # o un número entero >= número de clases
    stratify=y,
    random_state=42
)
```
