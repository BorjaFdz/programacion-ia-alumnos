# FUNCIONES



## ¿Qué es una función?
- Una o más líneas de código que funcionan como una unidad para realizar una tarea concreta.
- Las funciones pueden o no devolver una respuesta.
- Las funciones pueden o no tener parámetros.
- Cuando están dentro de una *clase*, también se les llama *métodos*.

## ¿Y para qué se utiliza una función?
- Para reutilizar el mismo código cuando sea de nuestro interés.
	- Dentro del mismo script.
	- Desde otros archivos.
- Nos aislan conceptualmente de lo que está sucediendo en el código.

````python
import random 
l_nombres = []
lista_nombres = ["   María"," Ainara"," Juan"," Daniel", " Miriam", " Ane"]
for nombre in lista_nombres:
    if nombre.endswith("a"):
        nombre = nombre.upper()
        nombre = nombre.strip()###############################################
        nombre = "_".join(nombre)#############################################
        nombre = nombre + ","+ str(random.randint(0,101))#####################
    else:
        nombre = nombre.lower()
        nombre = nombre.strip()###############################################
        nombre = "_".join(nombre)#############################################
        nombre = nombre + ","+ str(random.randint(0,101))#####################
    
    l_nombres.append(nombre)
print(l_nombres)
````

````python
import random 

def cambiar_nombre(name):
    """
    Esta función toma como parámetro un nombre de la lista_nombres y lo preprocesa.
    
    Parámetros:
    name str: el nombre de la persona
    
    Retorna:
    
    name str: nombre con espacios quitados, unidas las letras por "_" y añadido
              "," + un número aleatorio del 0 al 100.
    """
    
    name = name.strip()
    name = "_".join(name)
    name = name + ","+ str(random.randint(0,101))
    return name


l_nombres = []
lista_nombres = ["   María"," Ainara"," Juan"," Daniel", " Miriam", " Ane"]
for nombre in lista_nombres:
    if nombre.endswith("a"):
        nombre = nombre.upper()
        nombre = cambiar_nombre(nombre)
    else:
        nombre = nombre.lower()
        nombre = cambiar_nombre(nombre)
        
    l_nombres.append(nombre)
print(l_nombres)

````

## Parámetros y/o argumentos de una función
Realmente se refieren a lo mismo aunque desde la perspectiva de una función:
- Se llama argumento a los valores que se mandan a la función cuando esta es llamada.
- Se llama parámetros a los valores que recibe la función.

### Tipos de parámetros:
- **argumentos obligatorios o posicionales**
- **argumentos por palabra clave**

- __** kwargs__: argumentos arbitrarios por palabra clave. Le podemos dar el nombre que queramos mientras le precedan dos ** .
- argumentos con valor por defecto u opcionales.

````python
# *args: parámetros, posicionales arbitrarios
def coche(tipo,*args):
    """
    Parámetros:
        tipo: tipo de coche. SUV, Todoterreno, Berlina, Camioneta.
        *args: consideraremos una serie de parámetros arbitrarios relacionados con mantenimientos
    """
    print(f"Tipo de coche: {tipo}")
    print(type(args))
    print()
    print("##### Mantenimientos realizados: #####")
    for mantenimiento in args:
        print(f"\t - Tipo de acción: {mantenimiento}")
````

````python
# *args: parámetros, posicionales arbitrarios
def coche(tipo,*args,**kwargs):
    """
    Parámetros:
        tipo: tipo de coche. SUV, Todoterreno, Berlina, Camioneta.
        *args: consideraremos una serie de parámetros arbitrarios relacionados con mantenimientos
        **args: consideraremos una serie de parámetros arbitrarios por clave relacionados con propiedades del coche
    """
    print(f"Tipo de coche: {tipo}")
    print(type(args))
    print()
    print("##### Mantenimientos realizados: #####")
    for mantenimiento in args:
        print(f"\t - Tipo de acción: {mantenimiento}")
    propiedades = {}
    propiedades["tipo"] = tipo
    propiedades["otros"]  = []
    propiedades["propiedades_compra"]  = []
    print(type(kwargs))
    for clave, valor in kwargs.items():
        if clave == "matricula":
            propiedades["MATRICULA_VEHICULO"] = valor
        elif clave == "anio_compra":
            propiedades["propiedades_compra"].append({"Antigüedad del vehículo": datetime.datetime.now().year - int(valor)})
        elif clave == "importe":
            propiedades["propiedades_compra"].append({"Importe de compra del vehículo": str(valor) + " €"})
        else:
            propiedades["otros"].append({clave:valor})
    return propiedades
````
