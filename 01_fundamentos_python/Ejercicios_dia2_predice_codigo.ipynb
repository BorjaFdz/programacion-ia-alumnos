## 10. Igualdad, identidad y referencias

`==` pregunta si dos valores son equivalentes. `is` pregunta si son **el mismo objeto**.

### 10.1. `==` frente a `is`

> **Antes de ejecutar:** predice → justifica → ejecuta → corrige tu modelo mental.

**Conceptos:** igualdad · identidad · referencias



````python
a = [1, 2]
b = [1, 2]
c = a

print(a == b)
print(a is b)
print(a is c)
id(a) == id(b), id(a) == id(c)
````

<details>
    <summary><strong>Ver solución y explicación</strong></summary>


Salida:

```text
True
False
True
```

`a` y `b` tienen el mismo contenido, pero son listas distintas. `c` referencia exactamente el mismo objeto que `a`.


</details>

### 10.2. Modificar durante una iteración: razona antes de hacerlo

> **Antes de ejecutar:** predice → justifica → ejecuta → corrige tu modelo mental.

**Conceptos:** iteración · copia · mutabilidad



````python
datos = [1, 2, 3, 4]

for x in datos[:]:
    if x % 2 == 0:
        datos.remove(x)

print(datos)
````

<details>
    <summary><strong>Ver solución y explicación</strong></summary>


Salida:

```text
[1, 3]
```

Se recorre una **copia** (`datos[:]`) mientras se modifica la lista original. Esto evita muchos problemas típicos de modificar la colección que se está iterando.


</details>

> **Antes de ejecutar:** predice → justifica → ejecuta → corrige tu modelo mental.

````python
def normalizar_nombres(nombres):
    for i, nombre in enumerate(nombres):
        nombres[i] = nombre.strip().lower()
    return nombres

datos = [" Ana ", "BORJA", "  Python"]
resultado = normalizar_nombres(datos)

print(datos)
print(resultado)
print(datos is resultado)
id(datos), id(resultado)
````
