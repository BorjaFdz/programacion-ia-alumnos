```python
import random
# generar 10 numeros aleatorios entre 0 y 100
lista_aleatorios = []
for i in range(10):
    lista_aleatorios.append(random.randint(0,100))
    
lista_aleatorios
```

```python
# generar 10 numeros decimales entrero entre 0 y 1
lista_aleatorios = []
for i in range(10):
    lista_aleatorios.append(random.random()*100)
```

```python
# escoger 5 números aleatoriamente de un rango que vaya de 2.4 a 5.6
lista_aleatorios = []
for i in range(5000000):
    lista_aleatorios.append(random.uniform(2.4,266))
```

```python
random.choice(lista_aleatorios)
random.choices(lista_aleatorios, k=5)
random.sample(lista_aleatorios, k=5)
```

```python
random.seed(42)
lista = []
for i in range(1_000_000):
    lista.append(random.normalvariate(12,1))
    
import matplotlib.pyplot as plt

plt.figure(figsize=(20,10))
plt.hist(lista, bins = 1000)
plt.title("Mi súper distribución normal en forma de barras", fontsize=20)
plt.axvline(x=12, color='y', linestyle='--', linewidth=2, label='Media')
plt.axhline(y=25, color='r', linestyle='--')

    
```
