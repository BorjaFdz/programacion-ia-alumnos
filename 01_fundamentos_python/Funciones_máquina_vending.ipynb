```python
import numpy as np
```

```python
PRODUCTOS = np.array([
    ["cocacola","icetea","fanta"],
    ["kitkat","snats","pipas"],
    ["s_vegetal","s_cangrejo","s_tortilla"],
])

PRECIOS = np.array([
    [2, 2.5, 2],
    [1.9,2,0.85],
    [2.2, 2, 2.5],
])

CODIGOS = {
    "A0": [0,0],
    "A1": [0,1],
    "A2": [0,2],
    "B0": [1,0],
    "B1": [1,1],
    "B2": [1,2],
    "C0": [2,0],
    "C1": [2,1],
    "C2": [2,2],
}
```

```python
"A_A1" in CODIGOS
```

```python
def calcular_cambio(dinero_ingresado: float, dinero_pendiente: float):
    """
    Calcula el cambio a devolver tras ingresar dinero para un producto.

    Parameters
    ----------
    dinero_ingresado : float
        Cantidad total de dinero que ha introducido el usuario.
    dinero_pendiente : float
        Precio del producto que debe pagarse.

    Returns
    -------
    float
        Diferencia entre dinero_ingresado y dinero_pendiente.
        - Si es positiva: cantidad a devolver como cambio.
        - Si es cero: pago exacto.
        - Si es negativa: cantidad que aún falta por ingresar.
    """
    
    if dinero_pendiente == dinero_ingresado:
        pendiente = 0.0
    else:
        pendiente = dinero_ingresado - dinero_pendiente
    return pendiente

def comprar_producto(nombre="terrícola"):
    """
    Simula el proceso de compra de un producto en una máquina vending.
    Muestra un saludo, pide el código del producto, verifica que existe,
    solicita el importe necesario, gestiona pagos parciales si el usuario
    no ingresa suficiente dinero de una vez, y devuelve un mensaje con el
    producto y el cambio (si corresponde).

    Parameters
    ----------
    nombre : str, opcional
        Nombre o identificador del usuario. Por defecto es "terrícola".

    Returns
    -------
    str
        Mensaje final de la transacción:
        - Confirmación de entrega del producto y cambio (si hay).
        - Mensaje de error si el código no existe.

    Raises
    ------
    (Ninguna explícita, pero asume que están definidas globalmente:
     CODIGOS, PRECIOS, PRODUCTOS)
    """
    print(f"Hola {nombre}")
    codigo_producto = input("Inserte el código del producto deseado: ").strip().upper()
    print("Código introducido: ", codigo_producto)
    if codigo_producto in CODIGOS.keys():
        coordenadas_precio = CODIGOS[codigo_producto]
        dinero_a_ingresar = PRECIOS[coordenadas_precio[0]][coordenadas_precio[1]]
        dinero = float(input(f"Inserte {dinero_a_ingresar}€: "))
        while dinero <= 0:
            dinero = float(input(f"Introduzca una cantidad válida. Inserte {dinero_a_ingresar}€: "))
            
        cambio = calcular_cambio(dinero, dinero_a_ingresar)
        while cambio < 0:
            dinero_resto = round(float(input(f"Quedan {-cambio}€ por ingresar: ")), 1)
            cambio = calcular_cambio(dinero + dinero_resto, dinero_a_ingresar)
        if cambio > 0:
            mensaje = f"Aquí tiene su {PRODUCTOS[coordenadas_precio[0]][coordenadas_precio[1]]} y los {round(cambio, 1)} € de cambio. Gracias!"
        else:
            mensaje = f"Aquí tiene su {PRODUCTOS[coordenadas_precio[0]][coordenadas_precio[1]]}. Gracias!"
    else:
        mensaje = "El código no existe"
    return mensaje
```

```python
comprar_producto("borja")
```
