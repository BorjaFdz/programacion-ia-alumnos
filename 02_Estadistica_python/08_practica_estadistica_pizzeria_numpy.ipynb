## Ejercicio 9. Detectar outliers mediante IQR

Detecta valores atípicos de `ticket_medio_eur` usando la regla de Tukey:

- límite inferior = Q1 - 1.5·IQR
- límite superior = Q3 + 1.5·IQR

Muestra los clientes detectados.


<details>
<summary><strong>Mostrar solución</strong></summary>

```python
ticket = columna("ticket_medio_eur")
ids = columna("id_cliente")

q1, q3 = np.percentile(ticket, [25, 75])
iqr = q3 - q1
lim_inf = q1 - 1.5 * iqr
lim_sup = q3 + 1.5 * iqr

mascara = (ticket < lim_inf) | (ticket > lim_sup)

print("Límite inferior:", lim_inf)
print("Límite superior:", lim_sup)
print("Outliers detectados:")

for cliente_id, valor in zip(ids[mascara], ticket[mascara]):
    print(cliente_id, valor)
```
