```python
def suma(a,b): #funcion que toma a y b, las suma y devuelve el resultado de la suma
    """
    Description:
    funcion que toma a y b, las suma y devuelve el resultado de la suma
    Parameters:
        a: numero
        b: numero
    Result:
        a+b: numero
    """
    return a+b
```

```python
#Mejores prácticas para definir funciones. Pythonic Mode.
def suma(a: int | float, b: int | float) -> int | float:
    """Suma dos números.

    Args:
        a: Primer sumando.
        b: Segundo sumando.

    Returns:
        El resultado de sumar `a` y `b`.
    """
    return a + b
```

```python
resultado = suma(37,1)
```
