# Diccionario con 5 alimentos:
- por cada alimento: un diccionario con sus macros nutricionales
- una de las claves de cada alimento será un texto con una recomendación de ingesta.

```python
alimentos = {
  "yuca": {
    "energia": 160,
    "grasa": 0.3,
    "carbos": 38,
    "protes": 1.4,
    "recomendacion": "Úsala como fuente principal de carbohidratos en comidas pre/post entreno o en días de alta actividad. Controla la ración si buscas déficit (ej. 150–200 g cocida). Combínala con proteína (huevo, pollo, pescado) y vegetales para equilibrar el plato."
  },
  "tomate": {
    "energia": 18,
    "grasa": 0.2,
    "carbos": 3.9,
    "protes": 0.9,
    "recomendacion": "Ideal para añadir volumen y micronutrientes con muy pocas calorías. Úsalo libremente en ensaladas, guarniciones y salsas caseras. Excelente opción en déficit calórico para aumentar saciedad sin apenas impacto energético."
  },
  "huevo": {
    "energia": 155,
    "grasa": 10.6,
    "carbos": 1.1,
    "protes": 12.6,
    "recomendacion": "Fuente completa de proteína de alto valor biológico y grasas saludables. Muy útil para alcanzar proteína diaria (2–3 huevos en comidas principales). En déficit, puedes combinar 1 huevo entero + claras extra para mantener proteína y reducir grasa/calorías."
  }
}
```

# Crear 4 listas: energia, grasa, carbos, protes
- Recorrer el diccionario para meter los valores de cada alimento en cada lista.
- Con un bucle for

```python
alimentos.keys()
```

```python
alimentos.values()
```

```python
energia = []
carbos = []
grasa = []
protes = []

for alimento in alimentos:
    energia.append(alimentos[alimento]["energia"])
    carbos.append(alimentos[alimento]["carbos"])
    grasa.append(alimentos[alimento]["grasa"])
    protes.append(alimentos[alimento]["protes"])
```

```python
len(alimentos), len(energia)
```

```python
print(energia)
```

```python
max_e = max(energia)
min_e = min(energia)
print(f" los valores máaximos y mínimos de energia son: {max_e}, {min_e}")
max_c = max(carbos)
min_c = min(carbos)
print(f" los valores máaximos y mínimos de carbos son: {max_c}, {min_c}")
max_p = max(protes)
min_p = min(protes)
print(f" los valores máaximos y mínimos de protes son: {max_p}, {min_p}")
max_g = max(grasa)
min_g = min(grasa)
print(f" los valores máaximos y mínimos de grasa son: {max_g}, {min_g}")
```
