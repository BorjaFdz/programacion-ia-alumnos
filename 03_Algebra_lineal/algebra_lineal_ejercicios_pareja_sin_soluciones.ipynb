## Ejercicio 11 — Determinante de una matriz 🟥 (hay un bug)

**El caso real:** para "deshacer" una transformación (por ejemplo, pasar de un reporte ajustado con recargo/descuento otra vez al gasto original), se necesita poder **invertir** la matriz de transformación. Una matriz solo se puede invertir si su **determinante es distinto de 0**. Si el determinante es 0, la transformación "aplasta" la información y no hay vuelta atrás.

El código de abajo quiere comprobar si una matriz de transformación 2x2 es invertible, pero usa mal la función de NumPy. Encuentra el error y corrígelo.


```python
matriz_transformacion = np.array([[1.02, 0.0],
                                    [0.0, 0.95]])

determinante = np.linalg.det[matriz_transformacion]

if determinante != 0:
    print("Es invertible, determinante:", determinante)
else:
    print("No es invertible")
```
